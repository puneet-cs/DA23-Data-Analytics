# arange(), shuffle(), permutation(), linspace()

```python
# arange(start, end, step) : generate sequence of values from start to end-1 with given step size

arr1 = np.arange(1, 10)
arr1

# This is same as range() function in python
```

```python
arr2 = np.arange(1, 10, 2)
arr2
```

```python
# shuffle() : this will shuffle the elements

# shuffle() function perform the shuffling of data "Inplace"
# shuffle() function shuffle the elements in the same array (arr1).

np.random.shuffle(arr1)
arr1
```

```python
# permutation() : this will rearange the elements

# permutation() function don't change the actual array (arr1)

arr4 = np.random.permutation(arr1)
arr4
```

```python
# linspace() : linear space function generate the same difference elements

arr5 = np.linspace(1,10,4)
arr5
```

# 4. Array size & reshaping & transpose

```python
south_festive_dresses = np.array([["Lungi", "Dhoti", "Pancha"], 
                                  ["Sharee", "Sharara", "Half Saree"]])

south_festive_dresses
```

```python
south_festive_dresses.shape
```

```python
# reshape() : used to convert dimension of array
reshaped_arr = south_festive_dresses.reshape(6)
reshaped_arr
```

```python
# reshape to 3,2
reshaped_arr = south_festive_dresses.reshape(3,2)
reshaped_arr
```

```python
reshaped_arr = south_festive_dresses.reshape(3,2,1)
reshaped_arr
```

```python
# reshape() function can convert one type of array to another another array
# But make sure final array also have same number of elements
```

```python
# ravel() : convert any D array to 1D array

new_arr = south_festive_dresses.ravel()
print(new_arr)
```

```python
# transpose : row <-> column

south_festive_dresses
```
