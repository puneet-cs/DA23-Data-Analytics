# Copy() method

```python
# Def: copy() method is used to create a new independent copy/object, 
#             to avoid the accidental change in original data.

# Type of copy operations:
# 1. df1 = df  : Assignment : giving two names to same dataframe
# 2. deep copy : New duplicate copy of dataset
# 3. Shallow copy : New dataset structure but common shared data
```

```python
# Example -1 : Here df1 & df are just two different name for same data
df1 = df
df1
```

```python
# Example-2: Deep copy : This will create a separate copy 

fav_places = pd.DataFrame({
                    'Place' : ['ladakh', 'Keralam', 'Gachiboli', 'Kokapat'],
                    'Package' : [80, 60, 10, 15]
})

new_copy = fav_places.copy()
new_copy.iloc[0,0] = 'Dubai'
fav_places
```

```python
# Example-3 : Shallow copy : deep= False

# New dataset structure but common shared data

fav_places = pd.DataFrame({
                    'Place' : ['ladakh', 'Keralam', 'Gachiboli', 'Kokapat'],
                    'Package' : [80, 60, 10, 15]
})

new_copy = fav_places.copy(deep= False)
new_copy.iloc[0,0] = 'NewYork'
fav_places
```

# Merging two dataframes

```python
# merge() function can be used to combine two dataframes based upon common column
# merge() function works same as SQL Join operation : inner, outer, left, right

df1 = pd.DataFrame({'ID' : [1,2,3], 'Dream_company' : ['Amazon', 'Delloite', 'Google']})
df2 = pd.DataFrame({'ID' : [2,3,4], 'Dream_Package' : [25, 30, 40]})

# showing the inner merge : combine only common records
merged_df = pd.merge(df1, df2, on = 'ID', how = 'inner' )
merged_df
```

```python
merged_df = pd.merge(df1, df2, on = 'ID', how = 'outer' )
merged_df
```

```python
merged_df = pd.merge(df1, df2, on = 'ID', how = 'left' )
merged_df
```

```python
merged_df = pd.merge(df1, df2, on = 'ID', how = 'right' )
merged_df
```

# join() : combine two dataframe mainly based upon Index

```python
df3 = pd.DataFrame({'Name' : ['Chinky', 'Minky', 'Pinky']}, index = [1,2,3])
df4 = pd.DataFrame({'Salary' : [50, 100, 200]}, index = [2,3,4])

joined_df = df3.join(df4, how = 'inner')
joined_df
```

```python
joined_df = df3.join(df4, how = 'outer')
joined_df
```

```python
joined_df = df3.join(df4, how = 'left')
joined_df
```

```python
joined_df = df3.join(df4, how = 'right')
joined_df
```

# concat() : combine dataframe vertically or horizontally

```python
df5 = pd.DataFrame({"ID" : [1,2], "Movie" : ["Jersey", "offcampus"]})
df6 = pd.DataFrame({"ID" : [3,4], "Movie" : ["Jailer2", "Scene"]})

concat_df = pd.concat([df5, df6], axis = 0, ignore_index = True)
concat_df

# axis = 0 : vertically (row wise)
```

```python
# Horizontal concatination : Column wise
concat_df = pd.concat([df5, df6], axis = 1)
concat_df
```
