# 1. Data structures creation ( Series, DataFrame)

# Series : 1D array

```python
# Def: Series is 1D Homogeneous array in pandas.
```

```python
# creation of series : Empty series
ser1 = pd.Series()
print(ser1)
print(type(ser1))
```

```python
# example-2 
salary = [50.5, 40.6, 60.1, 70.0, 80, 100, "Hello"]

ser2 = pd.Series(salary)
print(ser2)
print(type(ser2))

# Series automatically add the indexing of elements
# in pandas strings are represented as object
```

```python
# Series of strings

Movies = ["Spiderman", "Puspa", "Devara", "Salaar", "Kalki"]

ser2 = pd.Series(Movies, index = ["A", "B", "C", "D", "E"])
print(ser2)
print(type(ser2))
```

```python
ser2[3]
```

```python
# Negative indexing is not allowed on pandas Series
ser2[-2]
```

```python
# create the pandas series from dictionary
d = {"Puri" : 40, "Dosa" : 60, "Idli" : 30}

ser2 = pd.Series(d)
print(ser2)
print(type(ser2))
```

```python
# convert the tuple into pandas series
tuple1 = (10, 20, 30, 40, 50)
ser3 = pd.Series(tuple1)
print(ser3)
```

```python
# creating single element pandas series
ser4 = pd.Series(100)
print(ser4)
```

# Update, delete, insert operations on Series

```python
ser5 = pd.Series([10, 20, 30, 40, 50])
print(ser5)
```

```python
# update operation
ser5[3] = 99
print(ser5)
```

```python
# insert operation
# if index don't exist then new element will be inserted
ser5[5] = 999
print(ser5)
```

```python
# we can insert the element with any new index value
# index value need not be in sequence
ser5[10] = 99999
print(ser5)
```

```python
ser5['A'] = 1000
print(ser5)
```

```python
ser5['A'] = 5000
print(ser5)
```

```python
# Delete element from Series
ser5 = ser5.drop(10)
print(ser5)
```

```python
# Multiple elements deletion
ser5 = ser5.drop([1,2,3,5])
print(ser5)
```

```python
ser5 = ser5.drop('A')
print(ser5)
```

# DataFrame : 2D array

```python
# Example-1 : creating empty dataframe

df = pd.DataFrame()
print(df)
```

```python
Reel_count = [20, 35, 1000, 40, 19, 100, 200, 300]

df = pd.DataFrame(Reel_count)
print(df)
print(type(df))
```

```python
import pandas as pd
```

```python
# Convert the dictionary to DataFrame
fruits_price = {"Seb": [150, 200, 300],"AragtiPandu": [70,80,90],"Mamidikaay": [80,90, 100]}

df = pd.DataFrame(fruits_price)
print(df)

# Key will become the column name
# Value will be inserting in that column
```

```python
# Convert a tuple of tuple into DataFrame

fruits_price = (("Seb", 150),("AragtiPandu", 70),("Mamidikaay", 80))
df = pd.DataFrame(fruits_price, columns = ["Fruit Name", "Price"])
print(df)
print(type(df))
```

```python
# convert a set of set into DataFrame
# We CANNOT store the set inside the set
# We cannot store the Mutable items inside the set
fruits_price = [("Seb", 150),("AragtiPandu", 70),("Mamidikaay", 80)]
df = pd.DataFrame(fruits_price, columns = ["Fruit Name", "Price"])
print(df)
print(type(df))
```

```python
# We can convert list of list into dataframe
# We can convert list of tuple into dataframe
# We can convert Tuple of tuple into Dataframe
# We cannot store set inside the set.
# We can convert set of tuple into datframe
```

```python
# Converting a Series into dataframe
ser1 = pd.Series([10, 20, 30, 40])

df = pd.DataFrame(ser1, columns = ["Values"])
print(df)
print(type(df))
```

```python
import os
os.getcwd()
```

```python
df = pd.read_excel("excel_data.xlsx")

# to_sql : create the table and write the df data 
df.to_sql("table_data", conn, if_exists = "replace")

# to_sql() function create the table "table_data"
# conn : is the connection with our database
# if_exists : replace table if already exist
```

```python
# Export data from table
exported_data = pd.read_sql("Select * From table_data", conn)
exported_data
```
